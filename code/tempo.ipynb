```python
from functools import wraps
def to_execute(result_key):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            iterator = func(*args, **kwargs)
            final_state = list(iterator)[-1]
            return getattr(final_state, result_key)
        return wrapper
    return decorator
```

```python
@to_execute('prod')
def lfact(n):
    return lproc(state=make_state({'n':n,'prod':1}),
                alive=lambda s: s.n > 1,
                update=lambda s: s._replace(n=s.n - 1, prod=s.prod*s.n))
```

```python
def make_counter(n):
    def f():
        nonlocal n
        n = (retval:= n) + 1
        return retval
    return f
```

```python
def make_random(init,a,b):
    import random
    def f():
        delta = random.randint(a,b)
        _init = init
        retval = _init
        _init = _init + delta
        return retval
    return f
```

```python
z = make_random(0,1,10)
```
