```python
def maxdepth(seq):
    """Find the maximum nesting level in seq
    In: list
    Out: int
    """
    if isinstance(seq, list):
        return 0 if not seq else max(maxdepth(seq[0]) + 1, maxdepth(seq[1:]))
    elif not isinstance(seq, list):
        return -1
```
