```python
from coursepy.math.peano import *
```

```python
def pred(a):
    assert peanop(a)
    if zerop(a):
        return zero()
    else:
        return a[0]
```

```python
def peano_sub(a, b):
    assert peanop(a) and peanop(b)
    if zerop(b) or zerop(a):
        return a
    else:
        return peano_sub(pred(a),pred(b))
```
